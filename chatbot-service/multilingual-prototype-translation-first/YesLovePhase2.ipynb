```python
import os
from dotenv import load_dotenv
from google import genai

load_dotenv()

client = genai.Client(api_key=os.getenv("GEMINI_API_KEY"))
```

```python
def detect_language(text):
    """
    Detects the language of the user's input.
    """
    return detect(text)
```

```python
class Translator:

    def __init__(self, provider="google"):
        self.provider = provider

    def translate(self, text, source, target):

        if self.provider == "google":
            return GoogleTranslator(
                source=source,
                target=target
            ).translate(text)

        # Future providers
        elif self.provider == "lelapa":
            pass

        elif self.provider == "bhala":
            pass
```

```python
translator = Translator()
```

```python
class LLMService:

    def __init__(self, provider="gemini"):
        self.provider = provider

    def generate(self, prompt):

        if self.provider == "gemini":

            response = client.models.generate_content(
                model="gemini-3.6-flash",
                contents=prompt
            )

            return response.text

        elif self.provider == "openai":
            # Future implementation
            pass

        elif self.provider == "claude":
            # Future implementation
            pass
```

```python
llm = LLMService(provider="gemini")
```

```python
def multilingual_chatbot():

    user_message = input("You: ")

    original_language = detect_language(user_message)

    print(f"Detected language: {original_language}")

    english_message = translator.translate(
        user_message,
        source="auto",
        target="en"
    )

    print(f"\nEnglish Translation:\n{english_message}")

    english_response = llm.generate(english_message)
    print(f"\nLLM Response:\n{english_response}")

    final_response = translator.translate(
        english_response,
        source="en",
        target=original_language
    )

    print(f"\nTranslated Response:\n{final_response}")
```

```python
multilingual_chatbot()
```

```python
def direct_multilingual_chatbot():

    user_message = input("You: ")

    response = llm.generate(user_message)

    print("\nLLM Response:\n")
    print(response)
```

```python
direct_multilingual_chatbot()
```

```python
import time

def run_direct(user_message):

    start = time.time()

    response = llm.generate(user_message)

    end = time.time()

    return {
        "response": response,
        "time": end - start
    }
```

```python
def run_pipeline(user_message):

    detected_language = detect_language(user_message)

    # Translate to English
    t1 = time.time()
    english = translator.translate(
        user_message,
        source="auto",
        target="en"
    )
    t2 = time.time()

    # LLM
    g1 = time.time()
    english_response = llm.generate(english)
    g2 = time.time()

    # Translate back
    b1 = time.time()
    final_response = translator.translate(
        english_response,
        source="en",
        target=detected_language
    )
    b2 = time.time()

    return {

        "language": detected_language,

        "english": english,

        "response": final_response,

        "translation_time": t2 - t1,

        "llm_time": g2 - g1,

        "back_translation_time": b2 - b1,

        "total_time":
            (t2 - t1) +
            (g2 - g1) +
            (b2 - b1)
    }
```

```python
def compare_architectures():

    user_message = input("Enter your message: ")

    direct = run_direct(user_message)

    pipeline = run_pipeline(user_message)

    print("\n==============================")

    print("USER MESSAGE")

    print(user_message)

    print("\n==============================")

    print("DIRECT GEMINI")

    print(direct["response"])

    print(f"\nTime: {direct['time']:.2f} seconds")

    print("\n==============================")

    print("TRANSLATION PIPELINE")

    print(pipeline["response"])

    print(f"\nDetected Language: {pipeline['language']}")

    print(f"Translation: {pipeline['translation_time']:.2f}s")

    print(f"LLM: {pipeline['llm_time']:.2f}s")

    print(f"Back Translation: {pipeline['back_translation_time']:.2f}s")

    print(f"Total: {pipeline['total_time']:.2f}s")

    print("\n==============================")

    if direct["time"] < pipeline["total_time"]:

        print("Direct LLM was faster.")

    else:

        print("Translation Pipeline was faster.")
```

```python
compare_architectures()
```
