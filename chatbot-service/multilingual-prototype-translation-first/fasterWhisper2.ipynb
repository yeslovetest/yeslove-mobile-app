```python
# ================================
# INSTALL LIBRARIES
# ================================
!pip -q install transformers torch soundfile faster-whisper sentencepiece

# ================================
# IMPORT LIBRARIES
# ================================
import torch
import soundfile as sf

from transformers import VitsModel, AutoTokenizer
from faster_whisper import WhisperModel

# ================================
# LOAD FASTER WHISPER
# ================================
print("Loading Faster Whisper...")

whisper = WhisperModel(
    "base",
    device="cpu",
    compute_type="int8"
)

print("Done!\n")

# ================================
# LOAD MMS-TTS
# ================================
print("Loading MMS-TTS model...")

model_name = "facebook/mms-tts-yor"

tokenizer = AutoTokenizer.from_pretrained(model_name)
tts = VitsModel.from_pretrained(model_name)

print("Done!\n")

# ================================
# TEXT TO SPEECH
# ================================
text = "Mo ń nímọ̀lára ìdánìkanwà lẹ́yìn ìpínya wa."

inputs = tokenizer(text, return_tensors="pt")

with torch.no_grad():
    waveform = tts(**inputs).waveform

sf.write(
    "yoruba.wav",
    waveform.squeeze().cpu().numpy(),
    16000
)


print("Audio generated!\n")

# ================================
# TRANSCRIBE WITH FASTER WHISPER
# ================================
segments, info = whisper.transcribe("yoruba.wav")

transcript = ""

for segment in segments:
    transcript += segment.text

print("====================================")
print("EXPECTED:")
print(text)

print("\nDETECTED LANGUAGE:")
print(info.language)

print("\nLANGUAGE CONFIDENCE:")
print(round(info.language_probability,3))

print("\nTRANSCRIPT:")
print(transcript)
print("====================================")
```
