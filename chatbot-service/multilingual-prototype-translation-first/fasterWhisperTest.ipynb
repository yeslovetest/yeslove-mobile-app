```python
from faster_whisper import WhisperModel
import time
```

```python
model = WhisperModel(
    "base",
    device="cpu",
    compute_type="int8"
)
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
segments, info = model.transcribe("WhatsApp Audio 2026-08-01 at 21.27.03.opus")
```

```python
print("Detected language:", info.language)
print("Language probability:", info.language_probability)
```

```python
transcript = ""

for segment in segments:
    transcript += segment.text

print(transcript)
```

```python
start = time.time()

segments, info = model.transcribe("WhatsApp Audio 2026-08-04 at 18.36.46.opus")

end = time.time()

print("Time:", round(end-start,2), "seconds")
```
